```python
from graphviz import Digraph

# Definir os códigos e suas descrições

codigos = {
    'Código1': 'Preparação e processamento de dados para gerar clusters com indicadores socioeconômicos.',
    'Código2': 'Exporta imagens de satélite correspondentes aos clusters usando a API do Google Earth Engine.',
    'Código3': 'Pré-processa arquivos TFRecord brutos exportados do Google Earth Engine.',
    'Código4': 'Cria divisões de dados (folds) para treinamento, validação e teste.',
    'Código5': 'Carrega e processa imagens de satélite e indicadores socioeconômicos, realiza análise exploratória de dados.',
    'Código6': 'Treina modelos de deep learning para estimar indicadores socioeconômicos.',
    'Código7': 'Usa features dos modelos treinados para treinar modelos de regressão e avaliar o desempenho.',
    'Código8': 'Realiza análise detalhada do desempenho dos modelos e gera visualizações.',
}

# Criar o grafo
dot = Digraph(comment='Fluxo Sequencial do Experimento', format='png')

# Especificar a direção do grafo para representar a ordem de execução
dot.attr(rankdir='LR')

# Adicionar nós para os códigos em sequência
for i, (nome_codigo, descricao) in enumerate(codigos.items(), start=1):
    label_codigo = f"{nome_codigo}\nOrdem: {i}\n{descricao}"
    dot.node(nome_codigo, label_codigo, shape='box', style='filled', fillcolor='lightyellow')

# Conectar os códigos em sequência
codigo_lista = list(codigos.keys())
for i in range(len(codigo_lista) - 1):
    dot.edge(codigo_lista[i], codigo_lista[i + 1], style='bold')

# Renderizar o grafo sem abrir o arquivo automaticamente
dot.render('fluxo_sequencial_experimento', format='png')
```
