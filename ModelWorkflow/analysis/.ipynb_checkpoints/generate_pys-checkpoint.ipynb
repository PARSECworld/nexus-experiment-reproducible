```python
import os
import subprocess

# Caminho para o diretório atual
diretorio = '.'

# Nome da subpasta onde os scripts serão salvos
subpasta = 'pys'

# Caminho completo para a subpasta
caminho_subpasta = os.path.join(diretorio, subpasta)

# Criar a subpasta se ela não existir
if not os.path.exists(caminho_subpasta):
    os.makedirs(caminho_subpasta)

# Percorrer os arquivos no diretório
for arquivo in os.listdir(diretorio):
    if arquivo.endswith('.ipynb') and arquivo.startswith('0'):
        # Comando para converter o notebook em script Python e salvar na subpasta
        comando = f"jupyter nbconvert --to script '{arquivo}' --output-dir '{caminho_subpasta}'"
        
        # Executar o comando
        subprocess.run(comando, shell=True, check=True)

        print(f"{arquivo} convertido para script Python em {caminho_subpasta}/.")
```
