```python
from diagrams import Diagram, Cluster, Edge
from diagrams.aws.storage import S3
from diagrams.custom import Custom
from diagrams.onprem.container import Docker
from diagrams.onprem.vcs import Git
from diagrams.onprem.workflow import Airflow
#from diagrams.aws.ml import SagemakerNotebookInstance
from diagrams.aws.ml import SagemakerNotebook

# Define the diagram
with Diagram("Deep Learning Reproducibility Architecture", show=False):
    # define icons
    mlflow_icon = Custom("MLflow", "./icons/mlflow.png")
    dvc_icon = Custom("DVC", "./icons/dvc.png")

    # define the components
    with Cluster("Data Management"):
        s3_data_lake = S3("Data Lake")
        dvc = dvc_icon

    with Cluster("Code Management"):
        git_repo = Git("Git Repository")
        docker = Docker("Docker Containers")

    with Cluster("Experiment Tracking"):
        mlflow = mlflow_icon

    with Cluster("Pipeline Orchestration"):
        airflow = Airflow("Apache Airflow")

    with Cluster("Cloud Services"):
        sagemaker = SagemakerNotebook("SageMaker Studio")

    # define interactions
    s3_data_lake - Edge(label="Stores") - dvc
    git_repo - Edge(label="Contains") - docker
    docker - Edge(label="Runs") - mlflow
    mlflow - Edge(label="Tracks Experiments") - airflow
    airflow - Edge(label="Orchestrates") - sagemaker

    s3_data_lake << Edge(label="Fetches") << mlflow
    sagemaker << Edge(label="Uses") << mlflow
```
