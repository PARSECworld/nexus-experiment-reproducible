```python
import boto3

# Parâmetros
vpc_id = 'vpc-737b890e'  # Substitua pelo ID do seu VPC
region = 'us-east-1'     # Substitua pela região correta

# Criar cliente EC2
ec2_client = boto3.client('ec2', region_name=region)

# Obter todos os grupos de segurança associados ao VPC
response = ec2_client.describe_security_groups(
    Filters=[
        {'Name': 'vpc-id', 'Values': [vpc_id]}
    ]
)

# Adicionar regra de entrada para a porta 8080 em todos os grupos de segurança
for sg in response['SecurityGroups']:
    sg_id = sg['GroupId']
    try:
        ec2_client.authorize_security_group_ingress(
            GroupId=sg_id,
            IpPermissions=[
                {
                    'IpProtocol': 'tcp',
                    'FromPort': 8080,
                    'ToPort': 8080,
                    'IpRanges': [{'CidrIp': '0.0.0.0/0'}]  # Ajuste conforme necessário
                }
            ]
        )
        print(f"Regra de segurança adicionada com sucesso ao grupo {sg_id}")
    except Exception as e:
        print(f"Erro ao adicionar regra ao grupo {sg_id}: {e}")

print("Modificação de grupos de segurança concluída.")
```
