```python
from collections import defaultdict
import os
import pickle
from pprint import pprint
import sys

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf

sys.path.append('/root/Datasets/SelectedClusters/')
from batchers import batcher, dataset_constants
from data_analysis.create_folds import create_folds, verify_folds
from utils.geo_plot import setup_ax, plot_locs
from utils.general import colordisplay, load_npz
from utils.plot import boxplot_df, plot_image_by_band
```

```python
indicators = ["income","literacy"]
for label in indicators:
    
    # Loading prediction values
    LOGS_ROOT_DIR = f'/root/logs/{label}/'
    model_dir = 'resnet_ms'
        
    test_path = os.path.join(LOGS_ROOT_DIR, model_dir, 'test_preds.npz')
    
    preds = load_npz(test_path)['test_preds']
    
    # Loading Folds data
    incountry_folds_path = '/root/Datasets/SelectedClusters/dhs_incountry_co.pkl'
    with open(incountry_folds_path, 'rb') as f:
        incountry_folds = pickle.load(f)
        
    # Loading Location of image
    
    file_path = f"/root/Datasets/SelectedClusters/dhs_co_{label}.npz"

    print(f"para o Indicador {label}:")
    
    result = load_npz(file_path)
    
    image_hists = result['image_hists']
    labels = result['labels']
    locs = result['locs']
    
    years = result['years']
    nls_center = result["nls_center"]

    nls_mean = result['nls_mean']
    
    labels = labels[incountry_folds['E']['test']]
    locs = locs[incountry_folds['E']['test']]
    preds = preds[incountry_folds['E']['test']]

    dmsp_mask = years < 2012
    viirs_mask = ~dmsp_mask
    _ = plot_locs(locs=locs, colors=labels, figsize=[12, 13],
              title=f'{label} Distribution', cbar_label=f'{label}', s = 5)
    plt.savefig(f'{label}_target.png')
    _ = plot_locs(locs=locs, colors=preds, figsize=[12, 13],
              title=f'{label} Predicted Distribution', cbar_label=f'{label} Predicted', s = 5)
    plt.savefig(f'{label}_preds.png')
    _ = plot_locs(locs=locs, colors= np.abs(labels - preds), figsize=[12, 13],
              title=f'{label} Difference Distribution', cbar_label=f'{label} (Target - Preds)', s = 5)
    plt.savefig(f'{label}_diff.png')
```

```python
def plot_splits(locs_dict, title=None):
    '''
    Args
    - locs_dict: dict, label (str) => np.array of lat, lon
    '''
    fig = plt.figure(figsize=[15, 15])
    ax = setup_ax(fig)

    for label, locs_array in locs_dict.items():
        ax.scatter(locs_array[:, 1], locs_array[:, 0], label=label, s=2)

    lgd = ax.legend()
    for handle in lgd.legendHandles:
        handle.set_sizes([50.0])

    if title is not None:
        ax.set_title(title)
    plt.show()
```

```python
label = 'income'
FOLDS = ['A','B','C','D','E']
# Loading Folds data
incountry_folds_path = '/root/Datasets/SelectedClusters/dhs_incountry_co.pkl'
with open(incountry_folds_path, 'rb') as f:
    incountry_folds = pickle.load(f)

# Loading Location of image
    
file_path = f"/root/Datasets/SelectedClusters/dhs_co_{label}.npz"

print(f"para o Indicador {label}:")
    
result = load_npz(file_path)
    
locs = result['locs']

locs_dict = {f: locs[incountry_folds[f]['test']] for f in FOLDS}
fig = plt.figure(figsize=[15, 15])
ax = setup_ax(fig)

for label, locs_array in locs_dict.items():
    ax.scatter(locs_array[:, 1], locs_array[:, 0], label=label, s=2)

lgd = ax.legend()
for handle in lgd.legendHandles:
    handle.set_sizes([50.0])

ax.set_title('In-country Folds')
plt.show()

fig.savefig("incountry_folds.jpg")
```

```python
def get_urban_rural_indices(locs, loc_dict):
    '''
    Args
    - locs: np.array, shape [N, 2]
    - loc_dict: dict, (lat, lon) => dict

    Returns
    - urban_indices: np.array, shape [num_urban]
    - rural_indices: np.array, shape [num_rural]
    '''
    urban_indices = []
    rural_indices = []

    for i, loc in enumerate(locs):
        loc = tuple(loc)
        if loc_dict[loc]['urban'] == 'RURAL':
            rural_indices.append(i)
        else:
            urban_indices.append(i)
    urban_indices = np.asarray(urban_indices)
    rural_indices = np.asarray(rural_indices)
    print(f'num urban: {len(urban_indices)}\t num rural: {len(rural_indices)}')
    return urban_indices, rural_indices

urban_indices, rural_indices = get_urban_rural_indices(locs, loc_dict)
```

```python
file_path = '/root/results.csv'
df = pd.read_csv(file_path)

df_train = df[df['split'] == "train_eval"]
df_val = df[df['split'] == "val"]

plt.figure(figsize = [10,5])
plt.subplot(121)
plt.suptitle("Mean Square Error for the ResNet-NL Backbone training",fontsize = 17)
plt.plot(range(151),df_train['mse'])
plt.title("Train Split")
plt.xlabel("Epochs")
plt.ylabel("mse")
plt.ylim([0.02,0.1])
plt.subplot(122)
plt.title("Validation Split")
plt.xlabel("Epochs")
plt.ylim([0.02,0.1])
plt.plot(range(151),df_val['mse'])
plt.savefig('mse_ResNet-NL.png')
```
