```
from graphviz import Digraph

def create_hierarchy_chart():
    dot = Digraph()

    # Root node
    dot.node("A", "Types of Solutions")

    # Level 1 nodes
    dot.node("B", "General Software Solutions")
    dot.node("C", "Niche Software Solutions")
    dot.node("D", "Machine Learning Solutions")
    dot.node("E", "Deep Learning Solutions")

    # Edges from root to level 1
    dot.edge("A", "B")
    dot.edge("A", "C")
    dot.edge("A", "D")
    dot.edge("A", "E")

    # Level 2 nodes for General Software Solutions
    dot.node("B1", "Generic Lifecycle Tools")
    dot.node("B2", "Metadata Models")
    dot.edge("B", "B1")
    dot.edge("B", "B2")

    # Level 2 nodes for Niche Software Solutions
    dot.node("C1", "Scientific Workflows")
    dot.node("C2", "Abstract Workflows")
    dot.edge("C", "C1")
    dot.edge("C", "C2")

    # Level 2 nodes for Machine Learning Solutions
    dot.node("D1", "Python Libraries")
    dot.node("D2", "ML Lifecycle Tools")
    dot.edge("D", "D1")
    dot.edge("D", "D2")

    # Level 2 nodes for Deep Learning Solutions
    dot.node("E1", "ML Lifecycle Tools")
    dot.node("E2", "Containers")
    dot.edge("E", "E1")
    dot.edge("E", "E2")

    # Save the diagram as a PNG file
    dot.render("hierarchy_chart", format="png", cleanup=True)
    print("Diagrama salvo como 'hierarchy_chart.png'")

# Generate the hierarchy chart
create_hierarchy_chart()
```
