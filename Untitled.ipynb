```
import mlflow

# Configura o MLflow para se conectar ao servidor remoto
mlflow.set_tracking_uri("http://3.235.42.129:5000")

# Inicia uma nova execução no MLflow
with mlflow.start_run(run_name="teste_simples"):
    # Registra uma métrica simples
    mlflow.log_metric("métrica_teste", 1.23)
    
    # Registra um parâmetro simples
    mlflow.log_param("parametro_teste", "valor_teste")
    
    # Finaliza a execução
    print("Execução registrada com sucesso no MLflow.")
```
